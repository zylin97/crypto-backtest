# BTC SMA均线交叉回测

使用7/25日均线交叉策略对BTC/USDT进行回测

```python
import sys
sys.path.append('..')

from backtest import BacktestEngine, load_ohlcv
from backtest.strategy import SMACross, BollingerBand
import matplotlib.pyplot as plt
```

```python
# 获取BTC/USDT日线数据
data = load_ohlcv('BTC/USDT', '1d', exchange_id='binance', limit=365)
print(f'数据范围: {data.index[0]} ~ {data.index[-1]}')
print(f'数据条数: {len(data)}')
data.tail()
```

```python
# SMA交叉策略回测
engine = BacktestEngine(data, initial_capital=10000)
result = engine.run(SMACross(fast_period=7, slow_period=25))

metrics = engine.get_metrics(result)
for k, v in metrics.items():
    print(f'{k}: {v}')
```

```python
# 权益曲线
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8), gridspec_kw={'height_ratios': [2, 1]})

ax1.plot(result['equity'], label='策略权益', color='#2196F3')
ax1.axhline(y=10000, color='gray', linestyle='--', alpha=0.5)
ax1.set_title('BTC SMA(7/25) 回测结果')
ax1.set_ylabel('权益 (USDT)')
ax1.legend()
ax1.grid(alpha=0.3)

ax2.plot(result['price'], label='BTC价格', color='#FF9800', alpha=0.7)
ax2.set_ylabel('价格 (USDT)')
ax2.legend()
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# 布林带策略对比
engine2 = BacktestEngine(data, initial_capital=10000)
result2 = engine2.run(BollingerBand(period=20, num_std=2.0))

print('\n=== 布林带策略 ===')
metrics2 = engine2.get_metrics(result2)
for k, v in metrics2.items():
    print(f'{k}: {v}')
```
